# 04. Executive Conclusions

This notebook turns the cleaned used-car analysis into a concise decision summary. It should be run after the first three notebooks.

```python
from pathlib import Path
import pandas as pd
from used_cars_analysis.analysis import executive_kpis, correlation_with_price, transmission_price_test, age_price_regression

project_root = Path.cwd()
while project_root != project_root.parent and not (project_root / 'pyproject.toml').exists():
    project_root = project_root.parent

clean_data_path = project_root / 'data' / 'cleaned_car_data.csv'
if not clean_data_path.exists():
    raise FileNotFoundError('Run 01_data_inspection.ipynb first to create data/cleaned_car_data.csv.')

df = pd.read_csv(clean_data_path)
kpis = executive_kpis(df)
kpis
```

## Market Signals

```python
price_correlations = correlation_with_price(df).drop('selling_price')
transmission_results = transmission_price_test(df)
regression_results = age_price_regression(df)

summary = pd.Series({
    'Strongest numeric price relationship': price_correlations.abs().idxmax(),
    'Automatic price premium': transmission_results['automatic_mean_price'] - transmission_results['manual_mean_price'],
    'Transmission test p-value': transmission_results['p_value'],
    'Price change per additional car year': regression_results['slope_car_age'],
    'Age regression R-squared': regression_results['r_squared'],
})
summary
```

## Interpretation

Use the direction and size of these signals to guide pricing and inventory decisions. Correlation and regression describe association, while the transmission test quantifies the observed difference between groups; neither establishes causation on its own.
